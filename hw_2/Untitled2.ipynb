```python
class Chocolate:
    def __init__(self, price: float, additional: str = None):
        if price <= 0:
            raise ValueError("Цена батончика должна быть больше нуля.")
        self.price = price
        self.additional = additional

    def show_my_chocolate(self):
        if self.additional:
            print(f"Шоколадный батончик, добавка: {self.additional}")
        else:
            print("Обычный шоколадный батончик")

    def __str__(self):
        return f"Шоколадный батончик, {self.price} рублей"


# пример
choc_1 = Chocolate(100.5)
choc_2 = Chocolate(100.9, "фундук")

# пример  метода show_my_chocolate
choc_1.show_my_chocolate()
choc_2.show_my_chocolate()


# Использование метода __str__ через print()
print(choc_1)  # Шоколадный батончик, 120.5 рублей
print(choc_2)  # Шоколадный батончик, 99.9 рублей


# Пример обработки ошибки при некорректной цене
try:
    choc_error = Chocolate(-10)
except ValueError as e:
    print(f"Ошибка: {e}")  # Ошибка: Цена батончика должна быть больше нуля.


```
