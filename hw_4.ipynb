```python
df = pd.read_csv('train.csv')
df.head()
```

```python
#смотрим значения в выживших -категориальный признак
df['Survived'].unique()
```

```python


plt.figure(figsize=(6, 4))
sns.countplot(data=df, x="Survived", palette=["#ff6b6b", "#4ecdc4"])
plt.xticks([0, 1], ["Не выжил (0)", "Выжил (1)"])
plt.title("Распределение Survived (количество)")
plt.ylabel("Количество пассажиров")
plt.xlabel("Survived")
plt.grid(axis="y", linestyle="--", alpha=0.3)
plt.tight_layout()
plt.show()
```

большая часть пассажирова не выжила


```python
#смотрим значения в Pclass -категориальный признак
df['Pclass'].unique()
```

```python
plt.figure(figsize=(6, 4))
ax = sns.countplot(data=df, x="Pclass", palette="muted")
new_labels = ["1й класс (1)", "2й класс (2)", "3й класс (3)"]
ax.set_xticklabels(new_labels)
plt.title("Распределение Pclass (количество)")
plt.ylabel("Количество пассажиров")
plt.xlabel("Pclass")
plt.grid(axis="y", linestyle="--", alpha=0.3)
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8, 5))
sns.histplot(data=df, x="Age", kde=True, bins=30, palette="pastel")
plt.title("Распределение возраста (Age)")
plt.xlabel("Возраст")
plt.ylabel("Количество пассажиров")
plt.grid(axis="y", linestyle="--", alpha=0.3)
plt.tight_layout()
plt.show()
```

Большая часть пассажиров - молодые люди в возрасте от 20 до 30 лет

```python
#смотрим значения в Sex -категориальный признак
df['Sex'].unique()
```

```python
plt.figure(figsize=(6, 4))
palette_map = {'male': '#4ecdc4',   # голубой для мужского
               'female': '#ff6b6b'}
sns.countplot(data=df, x="Sex", palette=palette_map)
plt.xticks(['male', 'female'], ["Мужской", "Женский"])
plt.title("Распределение Sex (количество)")
plt.ylabel("Количество пассажиров")
plt.xlabel("Sex")
plt.grid(axis="y", linestyle="--", alpha=0.3)
plt.tight_layout()
plt.show()
```

```python
#смотрим значения в Parch -категориальный признак
df['Parch'].unique()
```

```python

plt.figure(figsize=(7, 4))
ax = sns.countplot(data=df, x="Parch", palette="muted")

# Подписи: 0 → «без родителей/детей», 1 → «1 родственник» и т.д.
labels = [f"{i}" for i in range(len(ax.get_xticklabels()))]
ax.set_xticklabels(labels)

plt.title("Распределение по числу родителей/детей (Parch)")
plt.xlabel("Parch (число родителей/детей)")
plt.ylabel("Количество пассажиров")
plt.grid(axis="y", linestyle="--", alpha=0.3)
plt.tight_layout()
plt.show()
```

большая часть пассажиров плыли без сопровождения, одни

```python
#строим график box plot для age
plt.figure(figsize=(7, 4))
ax = sns.boxplot(data=df, x="Age")

plt.title("Распределение возраста (Age)")
plt.ylabel("Возраст")
plt.xlabel("")
plt.grid(axis="y", linestyle="--", alpha=0.3)
plt.tight_layout()
plt.show()
```

Основной возраст пассажиров от 20 до почти 40 лет не включая 40. Пассажиры в возрасте 65+ - выбросы. Медианный возраст ближе к 30 годам.  25 перцентиль - 20 лет, 75 перцинтиль около 40 лет. Есть небольшая ассиметрия распределения, так как медиана смещена к Q1. Границы нормально разброса от нуля до примерно 65 лет.

```python

counts = df["Survived"].value_counts()
labels = ["Не выжил (0)", "Выжил (1)"]
colors = ["#ff6b6b", "#4ecdc4"]

plt.figure(figsize=(7, 4))
plt.pie(
    counts,
    labels=labels,
    autopct="%1.1f%%",
    startangle=90,
    colors=colors
)
plt.title("Распределение Survived (%)")
plt.axis("equal")
plt.tight_layout()
plt.show()
```

```python
counts = df["Pclass"].value_counts()
labels = ["1й класс (1)", "2й класс (2)", "3й класс (3)"]
colors = ["#4ecdc4", "#ffe66d", "#ff6b6b"]

plt.figure(figsize=(7, 4))
plt.pie(
    counts,
    labels=labels,
    autopct="%1.1f%%",
    startangle=90,
    colors=colors
)
plt.title("Распределение по классам (%)")
plt.axis("equal")
plt.tight_layout()
plt.show()
```

```python
#посмотрим на вижиживших пассажиров в контектсет возраста, класса, стоимости билета

# удаляем пропуски
df_plot = df.dropna(subset=['Age', 'Fare'])

# Проверка: сколько строк осталось
print(f"Исходный размер: {len(df)}, после dropna: {len(df_plot)} ({len(df)-len(df_plot)} пропущено)")

# делаем категорию
df_plot['Pclass'] = df_plot['Pclass'].astype('category')

cols = ['Age', 'Fare', 'Pclass', 'Survived']

g = sns.pairplot(
    df_plot[cols],
    hue='Survived',
    diag_kind='kde',
    height=3,
    palette=['#ff6b6b', '#4ecdc4'],  # красный для погибших, бирюзовый для выживших
    plot_kws={'alpha': 0.6}  # добавляем прозрачность точкам, чтобы было видно наслоение
)

g.fig.suptitle('Pairplot: Age, Fare, Pclass (раскрашено по Survived)', y=1.02)
plt.show()

```

```python

import pandas as pd
import plotly.express as px
import plotly.io as pio
from google.colab import files
df_grouped = (
    df.groupby(['Pclass', 'Sex'])
      .size()
      .reset_index(name='count')
)

df_grouped['Pclass_label'] = df_grouped['Pclass'].map({
    1: '1-й класс',
    2: '2-й класс',
    3: '3-й класс'
})

df_grouped['Sex_label'] = df_grouped['Sex'].map({
    'male': 'Мужчины',
    'female': 'Женщины'
})
fig = px.sunburst(
    df_grouped,
    path=['Pclass_label', 'Sex_label'],
    values='count',
    color='Pclass_label',
    color_discrete_sequence=px.colors.qualitative.Set2,
    title='Распределение пассажиров: класс каюты → пол',
    branchvalues='total'
)

fig.update_traces(textinfo='label+percent entry')
fig.update_layout(margin=dict(t=40, l=0, r=0, b=0))
pio.renderers.default = "colab"
fig.show()
fig.write_html('Pclass_Sex.html')
```
