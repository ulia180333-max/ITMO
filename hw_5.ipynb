1. ЦЕЛОЧИСЛЕННЫЕ КОЛОНКИ (price, area, bedrooms, bathrooms, stories, parking):
   Все 6 колонок были int64 (8 байт). Предлагаю использовать следующие типы:
   - price  → uint32 (4 байта): диапазон 1.75–13.3 млн, uint32 покрывает до 4.3 млрд
   - area   → uint16 (2 байта): диапазон 1,650–16,200, uint16 покрывает до 65,535
   - bedrooms, bathrooms, stories, parking → uint8 (1 байт): значения 0–6
   
   Все значения неотрицательны, поэтому использованы беззнаковые типы (uint).

2. КАТЕГОРИАЛЬНЫЕ КОЛОНКИ (7 колонок типа object):
   - 6 бинарных (yes/no): mainroad, guestroom, basement, hotwaterheating,
     airconditioning, prefarea
   - 1 с 3 значениями (furnished/semi-furnished/unfurnished): furnishingstatus
   
   Предлагаю тип category, который хранит словарь уникальных значений
   + целочисленные коды (1 байт на элемент).



```python

# Диапазон: 1,750,000 – 13,300,000
# uint32 покрывает 0 – 4,294,967,295.
df_housing["price"] = df_housing["price"].astype("uint32")

# Диапазон: 1,650 – 16,200
# uint16 покрывает 0 – 65,535.
df_housing["area"] = df_housing["area"].astype("uint16")


# Диапазон: 1 – 6
# uint8 покрывает 0 – 255.
df_housing["bedrooms"] = df_housing["bedrooms"].astype("uint8")


# Диапазон: 1 – 4
# Аналогично bedrooms: uint8 с большим запасом.

df_housing["bathrooms"] = df_housing["bathrooms"].astype("uint8")


# Диапазон: 1 – 4
# Этажность — малое неотрицательное число.
df_housing["stories"] = df_housing["stories"].astype("uint8")


# Диапазон: 0 – 3
# Количество парковочных мест — малое неотрицательное целое.
df_housing["parking"] = df_housing["parking"].astype("uint8")

# --- mainroad, guestroom, basement, hotwaterheating,
#     airconditioning, prefarea: object → category ---
# Каждая колонка содержит 2 уникальных значения: "yes"/"no".
# category сохраняет исходные строковые метки для удобства.
for col in ["mainroad", "guestroom", "basement", "hotwaterheating",
            "airconditioning", "prefarea"]:
    df_housing[col] = df_housing[col].astype("category")

# --- furnishingstatus: object → category ---
# 3 уникальных значения: "furnished", "semi-furnished", "unfurnished".
# Category хранит 3 элемента в словаре + 1 байт на код для каждой строки.

df_housing["furnishingstatus"] = df_housing["furnishingstatus"].astype("category")



```

```python
# Смотрим текущие типы и потребление памяти после оптимизации
print("Типы данных после оптимизации:")
print(df_housing.dtypes)
print("\nПотребление памяти после оптимизации:")
print(df_housing.memory_usage(deep=True).sum() / 1024, "КБ")
```
